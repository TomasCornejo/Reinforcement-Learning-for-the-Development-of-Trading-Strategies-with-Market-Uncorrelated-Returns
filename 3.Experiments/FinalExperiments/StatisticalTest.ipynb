```python
PATH_DATA_LIST_METRICS_RESUME_TESTING  = glob(PREFIX_EXP + "/metrics_resume_testing.pkl")
```

```python
metrics_resume_testing_avg = last_data_list(PATH_DATA_LIST_METRICS_RESUME_TESTING)
```

```python
metrics_resume_testing_avg['DDPG Agent'].keys()
```

```python
metrica = 'net_worths'
metricDDPG = metrics_resume_testing_avg['DDPG Agent'][metrica]
metricSP500 = metrics_resume_testing_avg['SP500 Agent'][metrica]
```

```python
from scipy.stats import ttest_ind

t_stat, p_value = ttest_ind(metricDDPG, metricSP500, alternative='greater')
if p_value < 0.05:
    print("We reject the null hypothesis: Mean of DDPG Agent is significantly greater than Mean of SP500 Index.")
else:
    print("We fail to reject the null hypothesis: No significant difference in means.")
```
