```python
listResults = glob("OptimizeIterationsResults/*")
```

```python
path = "OptimizeIterationsResults\\OptunaDDPGExpBase_Total"
```

```python
path.split("\\")[-1]
```

```python
import pandas as pd
```

```python
def getHPParams(path):
    HPData = path + "\\trialsData.csv"
    df = pd.read_csv(HPData)
    df = df.sort_values(by="value",ascending=False)
    result_df = df[['value','params_batch_size','params_gamma','params_gradient_steps','params_learning_rate','params_tau','params_train_freq']].head(1)
    result_df['Agent'] = path.split("\\")[-1]
    return result_df
```

```python
filtered_list = listResults[2:]
```

```python
Finalsdfs = [getHPParams(path_in)  for path_in in filtered_list]
```

```python
finalParams = pd.concat(Finalsdfs, axis=0)
```

```python
finalParams.columns
```

```python
finalParams[['Agent','params_batch_size', 'params_gamma', 'params_gradient_steps', \
       'params_learning_rate', 'params_tau', 'params_train_freq']].to_csv('OptimizedParamsPerAgent.csv', index=False)
```
