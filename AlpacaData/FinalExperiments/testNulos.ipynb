```python
def getnulls(filename =''):
    df = pd.read_csv(filename)
    # filter the 
    filtered_df = df[df['timestamp'].str.contains('2018|2019|2020')]
    total = len(filtered_df)
    nulls = len(filtered_df[filtered_df['symbol'] == '0'])
    nullPercent = (nulls / total)
    return nullPercent
```
